```python
def reverse_list(lst):
    reversed_lst = []
    for i in range(len(lst)):
        reversed_lst.append(lst[len(lst) - 1 - i])
    return reversed_lst

print(reverse_list([1, 2, 3, 4, 5, 6]))
print(reverse_list(["A", "B", "C"]))
```

### 2

```python
def evans_and_odds(number):
    evens = 0
    odds = 0
    for i in range(number + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1
    print(f"Number of even numbers: {evens}")
    print(f"Number of odd numbers: {odds}")

print(evans_and_odds(100))
```

```python
def calculate_mean(numbers):
    total = 0
    for number in numbers:
        total += number
    mean = total / len(numbers)
    return mean

def calculate_median(numbers):
    numbers.sort()
    n = len(numbers)
    if n % 2 == 0:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2
    else:
        median = numbers[n // 2]
    return median

def calculate_mode(numbers):
    frequency = {}
    for number in numbers:
        if number in frequency:
            frequency[number] += 1
        else:
            frequency[number] = 1
    max_frequency = max(frequency.values())
    modes = [key for key, value in frequency.items() if value == max_frequency]
    return modes


```
