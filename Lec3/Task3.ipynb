### 1

```python
pi = 3.14

def area_of_circle(radius):
    area = pi * radius ** 2
    return area

def perimeter_of_circle(radius):
    perimeter = 2 * pi * radius
    return perimeter

print('Area of circle with radius 5:', area_of_circle(5))
print('Perimeter of circle with radius 5:', perimeter_of_circle(5))
```

### 2

```python
def add_all_nums(*nums):
    try:
        total = 0
        for num in nums:
            total += num
        return total
    except:
        return "Error: All arguments must be numbers"
    
print(add_all_nums(1, 2, 3, 4, 5))
print(add_all_nums(1, 2, '3', 4, 5))
```

### 3


```python
def convert_celsius_2_fahrenheit(celsius): 
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

print(convert_celsius_2_fahrenheit(100))
```

### 4

```python
def check_season(month):
    if month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    elif month in ['September', 'October', 'November']:
        return 'Autumn'
    else:
        return "Error: Invalid month"
    
print(check_season('April') )
print(check_season('December'))
```

### 5


```python
def calculate_slope(x1, y1, x2, y2):
    slope = (y2 - y1) / (x2 - x1)
    return slope

print(calculate_slope(0, 0, 1, 1))
```

### 6

```python
def solve_quadratic_eqn(a, b, c):
    root1 = (-b + (b**2 - 4*a*c)**0.5) / (2*a)
    root2 = (-b - (b**2 - 4*a*c)**0.5) / (2*a)
    roots = set([root1, root2])
    return roots

solve_quadratic_eqn(1, -3, 2)
```

### 8

```python
def capitalize_list_items(lst):
    for i in range(len(lst)):
        lst[i] = lst[i].capitalize()
    return lst

print(capitalize_list_items(['apple', 'banana', 'cherry']))
```

### 9

```python
def add_item(list, item):
    list.append(item)
    return list

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(add_item(food_staff, 'Eggs'))

numbers = [2, 3, 7, 9]
print(add_item(numbers, 6))
```

### 10

```python
def remove_item(list, item):
    if item in list:
        list.remove(item)
    return list

print(remove_item(food_staff, 'Mango'))
print(remove_item(numbers, 3))
```

### 11

```python
def sum_of_numbers(number):
    total = 0
    for i in range(number + 1):
        total += i
    return total

print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

### 12, 13

```python
def summ_of_odds(number):
    total = 0
    for i in range(number + 1):
        if i % 2 != 0:
            total += i
    return total

def summ_of_evens(number):
    total = 0
    for i in range(number + 1):
        if i % 2 == 0:
            total += i
    return total

print(summ_of_odds(10))
print(summ_of_evens(10))
```
