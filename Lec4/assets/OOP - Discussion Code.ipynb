### Question 1
Two separate objects from the same class, each calling `increment()` a different number of times. What are the final values?

```python
class Counter:
    def __init__(self):
        self.count = 0
    def increment(self):
        self.count += 1

c1 = Counter()
c2 = Counter()
c1.increment()
c1.increment()
c2.increment()

print(c1.count, c2.count)
```

### Question 3
This one is a classic trap. `members` is defined at the class level, not inside `__init__`. Two different `Team` objects each add one name. Are the lists separate or shared?

```python
class Team:
    members = []
    def __init__(self):
        self.members = []  # Instance Variable, unique to each instance of the class
    def add(self, name):
        self.members.append(name)

t1 = Team()
t2 = Team()
t1.add("Ali")
t2.add("Sara")
print(t1.members)
print(t2.members)
print(Team.members)
```

### Question 5
Two `Box` objects, each adding a different item, with `items=None` as the default. Are their lists independent this time?

```python
class Box:
    items = []
    def __init__(self, items=None):
        self.items = items if items is not None else []
    def add(self, item):
        self.items.append(item)

b1 = Box()
b2 = Box()
b3 = Box(["c"])
print(b1.items, b2.items, b3.items) #[],[],[c]
b1.add("a")         
b2.add("b") 

print(Box.items)            #[]

print(b1.items, b2.items, b3.items) #[a],[b],[c]
```

### Question 7
Two different `Point` objects with the same x and y values, compared with `==`. Does Python consider them equal?

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point(1, 2)

p2 = Point(1, 2)
print(p1 == p2)
print(p1 == p1)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __eq__(self, point):
       return self.x == point.x and self.y == point.y
   
p1 = Point(1, 2)
p2 = Point(1, 2)
print(p1 == p2)
print(p1 == p1)
```

```python
class Bucket:
    items = []
    def __init__(self):
        self.items = []

b1 = Bucket()
b2 = b1
b1.items.append("x")
b2.items.append("y")

print(b1.items, b2.items,Bucket.items)
```

### Question 15

### Question 17
`spend()` is a regular function (not a method), but it takes a `Wallet` object as an argument and modifies it directly. Does the change persist?

```python
class Wallet:
    def __init__(self, amount):
        self.amount = amount

def spend(wallet, amt):
    wallet.amount -= amt

w = Wallet(100)
spend(w, 30)
print(w.amount)
```
