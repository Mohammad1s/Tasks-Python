## Part 1

### 1

```python
try:
    age = int(input("Enter your age: "))
except:
    age = 0

if age >= 18 :
    print("You are old enough to drive")
else:
    diff = 18 - age
    print(f"You need {diff} more years to learn to drive")
```

### 2

```python
my_age = 21
your_age = int(input("Enter your age: "))

if my_age > your_age:
    differ = my_age - your_age
    if differ == 1:
        print(f"I'm {differ} year older than you") 
    else:
        print(f"I'm {differ} years older than you")   
    
elif my_age < your_age :
    differ = your_age - my_age
    if differ == 1:
        print(f"You're {differ} year older than me")
    else:
        print(f"You're {differ} years older than me")    
else:
    print("You are the same age as me!")
    
```

### 3

```python
month = input("Enter a month: ")

if month in ['September', 'October', 'November']:
    print("its Autumn")
elif month in ['December', 'January', 'February']:
    print("its Winter")
elif month in ['March', 'April', 'May']:
    print("its Spring")
elif month in ['June', 'July', 'August']:
    print("its Summer")
```
