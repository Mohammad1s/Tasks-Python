```python
import pandas as pd
import re
```

```python
df = pd.read_csv('nlp_customer_messages.csv', encoding='latin-1')
```

```python
print(df.head())
```

```python
Email = r"\w+@\w+\.\w+"
urls = r"https?://\S+|www\.\S+"
phones = r"010\d{8}"
numbers = r"\d+"



Emails = df['message'].str.findall(Email)
Urls = df['message'].str.findall(urls)
Phones = df['message'].str.findall(phones)
Numbers = df['message'].str.findall(numbers)

```

```python
print("numbers: ", Numbers)
```

```python
df["Emails"] = Emails
df["Urls"] = Urls
df["Phones"] = Phones
df["Numbers"] = Numbers.combine(
    Phones,
    lambda numbers, phones: [
        number for number in numbers if number not in phones
    ]
)
```

```python
for index, row in df.iterrows():
    for index, row in df.iterrows():
        df.at[index, 'message'] = re.sub(
            Email,
            'EMAIL',
            df.at[index, 'message']
        )
        df.at[index, 'message'] = re.sub(
            urls,
            'URL',
            df.at[index, 'message']
        )
        df.at[index, 'message'] = re.sub(
            phones,
            '(PHONE)',
            df.at[index, 'message']
        )
```
